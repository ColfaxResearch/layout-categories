# The category $\mathbf{Fact}$

A small companion to `tract.fact_morphism`. The category $\mathbf{Fact}$ has

- **objects**: flat tuples of positive integers, and
- **morphisms**: a morphism $F: (u_1,\ldots,u_p) \to (t_1,\ldots,t_n)$ is a tuple $(F_1,\ldots,F_n)$ where each $F_i$ is a flat tuple of positive integers with $\prod F_i = t_i$, and the concatenation of $F_1,\ldots,F_n$ equals $(u_1,\ldots,u_p)$.

In other words, a morphism partitions its domain into consecutive blocks, one per codomain entry, with each block multiplying to that entry. This is precisely a *flat refinement* $(u_1,\ldots,u_p) \twoheadrightarrow (t_1,\ldots,t_n)$ recorded by its relative modes (§2.2.4 of the paper).

```python
from tract import Fact_morphism, NestedTuple
```

## Constructing morphisms

A morphism is specified by its domain, codomain, and modes $(F_1,\ldots,F_n)$; the constructor validates the two defining conditions. Note that morphisms carry genuine data: distinct block structures give distinct morphisms between the same objects, so $\mathbf{Fact}$ is a category, not merely the refinement poset.

```python
F = Fact_morphism(domain=(2, 1, 2), codomain=(2, 2), modes=((2,), (1, 2)))
G = Fact_morphism(domain=(2, 1, 2), codomain=(2, 2), modes=((2, 1), (2,)))
print(F)
print(G)
print("F == G:", F == G)
```

```python
# Invalid data is rejected
try:
    Fact_morphism(domain=(2, 3), codomain=(5,), modes=((2, 3),))
except ValueError as e:
    print("ValueError:", e)
```

## Identity and composition

The identity on $(t_1,\ldots,t_n)$ is $((t_1),\ldots,(t_n))$. Composition concatenates blocks of blocks: as with `Tuple_morphism`, `f.compose(g)` is the diagrammatic composite $g \circ f$.

```python
f = Fact_morphism((2, 2, 3, 5), (4, 15), ((2, 2), (3, 5)))   # (2,2,3,5) -> (4,15)
g = Fact_morphism((4, 15), (60,), ((4, 15),))                 # (4,15)   -> (60,)
print("g ∘ f =", f.compose(g))

T = (4, 15)
print("identity on T:", Fact_morphism.identity(T))
print("unit laws hold:",
      f.compose(Fact_morphism.identity(f.codomain)) == f
      and Fact_morphism.identity(f.domain).compose(f) == f)
```

## The sum $\oplus$

$f \oplus g$ concatenates domains, codomains, and modes, making $\mathbf{Fact}$ monoidal; it satisfies the interchange law with composition.

```python
print("f ⊕ g =", f.sum(g))
```

## Relation to refinements

The modes of a morphism assemble into a nested tuple that refines the codomain, and every flat refinement arises this way — `from_refinement` inverts `refined_codomain`.

```python
R = F.refined_codomain()
print("modes as a nested tuple:", R)
print("refines the codomain:", R.refines(NestedTuple(F.codomain)))
print("round-trip:", Fact_morphism.from_refinement(R, NestedTuple(F.codomain)) == F)
```

## Remark on naming

Forgetting the integer values, a morphism $(u_1,\ldots,u_p) \to (t_1,\ldots,t_n)$ is exactly an *active* (endpoint-preserving) map $[p] \to [n]$ in the simplex category $\Delta$, so $\mathbf{Fact}$ is the category of elements of the product-pushforward functor on $\Delta_{\mathrm{act}}$ over the monoid $(\mathbb{Z}_{>0}, \times)$. No canonical name for this category appears in the literature; `Fact` (for *factorization*) is our working name.
